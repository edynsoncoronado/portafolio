# Calculating the cost

[Doc](https://developers.openai.com/api/docs/guides/token-counting)

## Method 1

```python
prompt = """
Who is Edynson Coronado?
"""

max_completion_tokens = 500

response_text = client.responses.create(
    model="gpt-5.4-nano",
    input=prompt,
    max_output_tokens=max_completion_tokens
)

print(response_text.output_text)
```

```python
print(f"Input: {response.usage.input_tokens}")
print(f"Output: {response.usage.output_tokens}")
print(f"Total: {response.usage.total_tokens}")
```

```python
INPUT_PRICE = 0.15 / 1_000_000  # $0.15 per 1M tokens
OUTPUT_PRICE = 0.60 / 1_000_000  # $0.60 per 1M tokens

input_cost = response.usage.input_tokens * INPUT_PRICE
output_cost = response.usage.output_tokens * OUTPUT_PRICE
total_cost = input_cost + output_cost

print(f"Costo input : ${input_cost:.8f}")
print(f"Costo output: ${output_cost:.8f}")
print(f"Costo total : ${total_cost:.8f}")
```

## Method 2

```python
response_text2 = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {"role": "user", "content": prompt}
    ],
    max_tokens=max_completion_tokens
)
```

```python
print(response_text2.choices[0].message.content)
```

```python
input_token_price = 0.15 / 1_000_000  # $0.15 per 1M tokens
output_token_price = 0.60 / 1_000_000  # $0.60 per 1M tokens

input_tokens = response_text2.usage.prompt_tokens  # Number of input tokens used
output_tokens = response_text2.usage.completion_tokens  # Number of output tokens used

print(f"Input tokens: {input_tokens}, Output tokens: {output_tokens}")

total_cost = (input_tokens * input_token_price) + (output_tokens * output_token_price)
print(f"Total cost: ${total_cost:.8f}")
```
