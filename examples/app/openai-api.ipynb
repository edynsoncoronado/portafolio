# Calculating the cost

[Doc](https://developers.openai.com/api/docs/guides/token-counting)

## Method 1

```python
prompt = """
Who is Edynson Coronado?
"""

max_completion_tokens = 500

response_text = client.responses.create(
    model="gpt-5.4-nano",
    input=prompt,
    max_output_tokens=max_completion_tokens
)

print(response_text.output_text)
```

```python
print(f"Input: {response.usage.input_tokens}")
print(f"Output: {response.usage.output_tokens}")
print(f"Total: {response.usage.total_tokens}")
```

```python
INPUT_PRICE = 0.15 / 1_000_000  # $0.15 per 1M tokens
OUTPUT_PRICE = 0.60 / 1_000_000  # $0.60 per 1M tokens

input_cost = response.usage.input_tokens * INPUT_PRICE
output_cost = response.usage.output_tokens * OUTPUT_PRICE
total_cost = input_cost + output_cost

print(f"Costo input : ${input_cost:.8f}")
print(f"Costo output: ${output_cost:.8f}")
print(f"Costo total : ${total_cost:.8f}")
```

## Method 2

```python
response_text2 = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {"role": "user", "content": prompt}
    ],
    max_tokens=max_completion_tokens
)
```

```python
print(response_text2.choices[0].message.content)
```

```python
input_token_price = 0.15 / 1_000_000  # $0.15 per 1M tokens
output_token_price = 0.60 / 1_000_000  # $0.60 per 1M tokens

input_tokens = response_text2.usage.prompt_tokens  # Number of input tokens used
output_tokens = response_text2.usage.completion_tokens  # Number of output tokens used

print(f"Input tokens: {input_tokens}, Output tokens: {output_tokens}")

total_cost = (input_tokens * input_token_price) + (output_tokens * output_token_price)
print(f"Total cost: ${total_cost:.8f}")
```

# Text Summarization

```python
from pathlib import Path

content = Path("Text MCP.md").read_text(encoding="utf-8")
```

```python
prompt = f"""Sumarize the following text in Spanish, keeping the main ideas and key points intact.
The summary should be concise, clear, and easy to understand.
Avoid adding personal opinions or interpretations. Here is the text:
{content}
"""

response = client.chat.completions.create(
  model="gpt-4o-mini",
  messages=[{"role": "user", "content": prompt}],
  max_completion_tokens=400
)

print(response.choices[0].message.content)
```

```python
input_token_price = 0.15 / 1_000_000
output_token_price = 0.6 / 1_000_000

# Extract token usage
input_tokens = response.usage.prompt_tokens
print(f"Input tokens: {input_tokens}")
output_tokens = response.usage.completion_tokens
print(f"Output tokens: {output_tokens}")
# Calculate cost
cost = (input_tokens * input_token_price) + (output_tokens * output_token_price)
print(f"Estimated cost: ${cost:.8f}")
```

# Shot prompting

```python
# Define a multi-line prompt to classify sentiment
prompt = """Classify sentiment as 1-5 (bad-good) using review from an online shoe:
1. Unbelievably good!
2. Shoes fell apart on the second use.
3. The shoes look nice, but they aren't very comfortable.
4. Can't wait to show them off!"""

# Create a request to the Chat Completions endpoint
response = client.chat.completions.create(
  model="gpt-4o-mini",
  messages=[{"role": "user", "content": prompt}],
  max_completion_tokens=100
)

print(response.choices[0].message.content)
```

```python
# Add the example to the prompt
prompt = """Classify sentiment as 1-5 (negative to positive):
1. Love these! = 5
2. Unbelievably good! =
3. Shoes fell apart on the second use. =
4. The shoes look nice, but they aren't very comfortable. =
5. Can't wait to show them off! ="""

response = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[{"role": "user", "content": prompt}],
    max_completion_tokens=100
)
print(response.choices[0].message.content)
```

```python
# Add the final example
prompt = """Classify sentiment as 1-5 (negative to positive):
1. Comfortable, but not very pretty = 2
2. Love these! = 5
3. Unbelievably good! = 
4. Shoes fell apart on the second use. = 
5. The shoes look nice, but they aren't very comfortable. = 
6. Can't wait to show them off! = """

response = client.chat.completions.create(
    model="gpt-4o-mini", 
    messages=[{"role": "user", "content": prompt}], 
    max_completion_tokens=100
)
print(response.choices[0].message.content)
```

# Guardrails

```python
sys_msg = """You are a study planning assistant that creates plans for learning new skills.

If these skills are non related to languages, return the message:

'Apologies, to focus on languages, we no longer create learning plans on other topics.'
"""
```

```python
user1_msg = """I want to learn how to play the guitar. Can you create a study plan for me?"""
```

```python

# Create a request to the Chat Completions endpoint
response = client.chat.completions.create(
  model="gpt-4o-mini",
  messages=[
    {"role": "system", "content": sys_msg},
    {"role": "user", "content": user1_msg}
  ]
)

print(response.choices[0].message.content)
```
